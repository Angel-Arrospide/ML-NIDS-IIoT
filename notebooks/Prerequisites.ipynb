```python
import os
import sys
from dotenv import load_dotenv

# Añadir el directorio raíz al path para poder importar src
root_path = os.path.abspath(os.path.join(os.getcwd(), '..'))
if root_path not in sys.path:
    sys.path.append(root_path)

from src.config import DATASETS

# Cargar variables de entorno (.env)
load_dotenv(os.path.join(root_path, '.env'))

print(f"Raíz del proyecto: {root_path}")
```

```python
import shutil
import kagglehub

for name, config in DATASETS.items():
    print(f"\n--- Procesando dataset: {name} ---\n")
    
    try:
        # 1. Descarga a la caché de kagglehub
        print(f"Descargando desde: {config['url']}")
        path_cache = kagglehub.dataset_download(config['url'])
        
        # 2. Copiar a la carpeta raw del proyecto
        target_path = config['raw_path']
        print(f"Copiando archivos a: {target_path}")
        
        # Usamos dirs_exist_ok=True para permitir ejecuciones repetidas
        shutil.copytree(path_cache, target_path, dirs_exist_ok=True)
        
        print(f"{name} listo.")
        
    except Exception as e:
        print(f"Error procesando {name}: {e}")
```

```python
for name, config in DATASETS.items():
    print(f"\n--- Directory tree for: {name} ---\n")
    path = config['raw_path']
    for dirpath, dirnames, filenames in os.walk(path):
        level = dirpath.replace(path, '').count(os.sep)
        indent = ' ' * 4 * (level)
        print(f'{indent}{os.path.basename(dirpath)}/')
        subindent = ' ' * 4 * (level + 1)
        for f in filenames:
            print(f'{subindent}{f}')
```
