```python
%load_ext autoreload
%autoreload 2

from src.config import DATASETS, BASE_PATH
from rich.panel import Panel

print(
    f"Raíz del proyecto: {BASE_PATH}\n",
    f"Datasets configurados: {', '.join(DATASETS.keys())}"
)
```

```python
import shutil
import kagglehub

for name, config in DATASETS.items():
    try:
        # 1. Descarga a la caché de kagglehub
        print(f"Descargando desde Kaggle: {config['url']}")
        path_cache = kagglehub.dataset_download(config['url'])
        
        # 2. Copiar a la carpeta raw del proyecto (usando Pathlib de config.py)
        target_path = config['raw_path']
        print(f"Sincronizando archivos en: {target_path}")
        
        # Aseguramos que la carpeta destino existe
        target_path.mkdir(parents=True, exist_ok=True)
        
        # Copiamos el contenido
        shutil.copytree(path_cache, target_path, dirs_exist_ok=True)
        
        print(f" + {name} procesado correctamente.")
        
    except Exception as e:
        print(f" - Error procesando {name}: {e}")
```
