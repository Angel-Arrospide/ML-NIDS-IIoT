```python
%load_ext autoreload
%autoreload 2

from src.config import DATASETS, console, BASE_PATH
from rich.panel import Panel

# Mostramos información del entorno de forma elegante
console.print(Panel(
    f"[bold info]Entorno de Trabajo Listo[/bold info]\n"
    f"Raíz del proyecto: [path]{BASE_PATH}[/path]\n"
    f"Datasets configurados: [yellow]{', '.join(DATASETS.keys())}[/yellow]",
    expand=False,
    border_style="blue"
))
```

```python
import shutil
import kagglehub
from rich.progress import Progress

for name, config in DATASETS.items():
    with console.status(f"[bold cyan]Procesando dataset: {name}...[/bold cyan]") as status:
        try:
            # 1. Descarga a la caché de kagglehub
            console.log(f"Descargando desde Kaggle: [underline]{config['url']}[/underline]")
            path_cache = kagglehub.dataset_download(config['url'])
            
            # 2. Copiar a la carpeta raw del proyecto (usando Pathlib de config.py)
            target_path = config['raw_path']
            console.log(f"Sincronizando archivos en: [path]{target_path}[/path]")
            
            # Aseguramos que la carpeta destino existe
            target_path.mkdir(parents=True, exist_ok=True)
            
            # Copiamos el contenido
            shutil.copytree(path_cache, target_path, dirs_exist_ok=True)
            
            console.print(f"[bold success]✓ {name} procesado correctamente.[/bold success]")
            
        except Exception as e:
            console.print(f"[bold error]✗ Error procesando {name}: {e}[/bold error]")
```

```python
from rich.tree import Tree
from pathlib import Path

def build_directory_tree(path: Path, tree: Tree):
    """Función recursiva para construir el árbol visual."""
    for item in sorted(path.iterdir()):
        # Ignoramos archivos ocultos
        if item.name.startswith('.'): continue
        
        if item.is_dir():
            branch = tree.add(f"[bold blue]📂 {item.name}/[/bold blue]")
            build_directory_tree(item, branch)
        else:
            tree.add(f"[green]📄 {item.name}[/green]")

for name, config in DATASETS.items():
    path = Path(config['raw_path'])
    
    if path.exists():
        console.print(f"\n[bold info]Estructura de archivos para {name}:[/bold info]")
        # Creamos la raíz del árbol
        tree = Tree(f"[bold cyan]{path.name}/[/bold cyan]")
        build_directory_tree(path, tree)
        console.print(tree)
    else:
        console.print(f"[warning]El path para {name} no existe todavía.[/warning]")
```
